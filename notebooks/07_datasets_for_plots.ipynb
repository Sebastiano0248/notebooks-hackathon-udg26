### 1r gràfic: CARAC CENSUS TRACT

```python
df_for_rent_synthetic = pd.read_csv(f"{PROCESSED_DATA}/girona_for_rent_synthetic.csv")
df_for_rent = pd.read_csv(f"{PROCESSED_DATA}/girona_for_rent.csv")

df_all_for_rent = pd.concat(
    [
        df_for_rent_synthetic.drop(columns=["synthetic_id", "year_available"]),
        df_for_rent,
    ],
    ignore_index=True,
)
df_all_for_rent["elevator"] = df_all_for_rent["elevator"].apply(lambda x: 1 if x else 0)
df_all_for_rent.drop(columns=["address", "neighbourhood"], inplace=True) # No es pot fer la mitja de columnes discretes
df_all_for_rent_grouped = df_all_for_rent.groupby('census_tract').mean().reset_index()
print(f"Is census_tract unique? {df_all_for_rent_grouped['census_tract'].is_unique}")
df_all_for_rent_grouped.head(5)
```

```python
df_all_for_rent_grouped.to_csv(f"{OBSERVABLE_DATA}/df_all_for_rent_grouped.csv", index=False)
```
